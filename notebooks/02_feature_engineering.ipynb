```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split, cross_val_score
import joblib
```

```python
data = pd.read_csv('D:/Profile Builder Projects/Student Progress Tracker/data/raw/Student_Performance.csv')
data.drop(data.select_dtypes(include='object'), axis=1, inplace=True)
data.drop(['student_id','age'], axis=1, inplace=True)
```

```python
data.fillna(data.select_dtypes(include='number').mean(), inplace=True)
print(data.head())
```

```python
y = data['overall_score']
X = data.drop('overall_score', axis=1)
print(X.head())
print(y.head())
```

```python
scaler = StandardScaler()
scaled = scaler.fit_transform(X)
scaled_df = pd.DataFrame(scaled, columns=X.columns)
scaled_df = pd.concat([scaled_df, y], axis=1)
scaled_df.to_csv('D:/Profile Builder Projects/Student Progress Tracker/data/processed/scaled_student_performance.csv', index=False)
```

```python
joblib.dump(scaler, 'D:/Profile Builder Projects/Student Progress Tracker/models/scaler.pkl')
```

```python
print(scaled_df.head())
```

```python
y = scaled_df['overall_score']
X = scaled_df.drop('overall_score', axis=1)
```

```python
X_train , X_test , y_train , y_test = train_test_split(X, y, test_size=0.2, random_state=40)
```

```python
train_data = pd.concat([X_train, y_train], axis=1)
test_data = pd.concat([X_test, y_test], axis=1)

train_data.to_csv('D:/Profile Builder Projects/Student Progress Tracker/data/processed/train_data.csv', index=False)
test_data.to_csv('D:/Profile Builder Projects/Student Progress Tracker/data/processed/test_data.csv', index=False)

trained_data = pd.read_csv('D:/Profile Builder Projects/Student Progress Tracker/data/processed/train_data.csv')
print(trained_data.head())

tested_data = pd.read_csv('D:/Profile Builder Projects/Student Progress Tracker/data/processed/test_data.csv')
print(tested_data.head())
```
