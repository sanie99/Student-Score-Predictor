```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error , r2_score
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
```

```python
train_data = pd.read_csv('../data/processed/train_data.csv')
test_data = pd.read_csv('../data/processed/test_data.csv')
```

```python
print(train_data.head())
print(test_data.head())
```

```python
# Separate features and target variable

X_train = train_data.drop('overall_score', axis=1)
y_train = train_data['overall_score']

X_test = test_data.drop('overall_score', axis=1)
y_test = test_data['overall_score']
```

```python
# Train the linear regression model

linear_model = LinearRegression()
linear_model.fit(X_train, y_train)
y_pred = linear_model.predict(X_test)
```

```python
import joblib
```

```python
joblib.dump(linear_model, '../models/linear_model.pkl')
```

```python
# Train the Support Vector Regression (SVR) model

svr_model = SVR(kernel='rbf')
svr_model.fit(X_train, y_train)
y_pred_svr = svr_model.predict(X_test)
```

```python
joblib.dump(svr_model, '../models/svr_model.pkl')
```

```python
# Train the Random Forest Regressor model

rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)
y_pred_rf = rf_model.predict(X_test)
```

```python
joblib.dump(rf_model, '../models/rf_model.pkl')
```
