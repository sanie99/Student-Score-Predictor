```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
```

```python
data = pd.read_csv('D:/Profile Builder Projects/Student Progress Tracker/data/raw/Student_Performance.csv')
```

```python
data.reset_index(inplace=True)
```

```python
print(data.info())
# A concise summary of the DataFrame, including the number of non-null entries and data types for each column.
```

```python
data = pd.DataFrame(data)
print(data.describe())
```

```python
print(data.head())
# Top 5 rows of the dataset to get a sense of the data structure and values.
```

```python
plt.figsize=(15,15)
sns.heatmap(data.corr(numeric_only=True), cmap="Blues",annot=True)
# Clearly shows that there is a strong positive correlation between overall_score and study_hours, and a moderate positive correlation between overall_score and attendance_rate.
```

```python
sns.pairplot(data)
# Visualizes the relationships between all pairs of numeric variables, helping to identify patterns and correlations in the data.
```

```python
plt.plot(data['overall_score'], data['study_hours'],'o',markersize=1)
plt.xlabel('Overall Score')
plt.ylabel('Study Hours')
plt.show()
```

```python
plt.hist(data['overall_score'], bins=20, edgecolor='black')
plt.title('Distribution of Overall Scores')
# The histogram shows that the overall scores are approximately normally distributed, with a slight skew towards higher scores, indicating that most students performed well, but there are some lower scores as well.
```
