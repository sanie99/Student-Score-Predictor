```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
import joblib
```

```python
train_data = pd.read_csv('../data/processed/train_data.csv')
test_data = pd.read_csv('../data/processed/test_data.csv')
```

```python
X_train = train_data.drop('overall_score', axis=1)
y_train = train_data['overall_score']

X_test = test_data.drop('overall_score', axis=1)
y_test = test_data['overall_score']
```

```python
# Loading the models

linear_model = joblib.load('../models/linear_model.pkl')
svr_model = joblib.load('../models/svr_model.pkl')
rf_model = joblib.load('../models/rf_model.pkl')
```

```python
linear_y_pred = linear_model.predict(X_test)
svr_y_pred = svr_model.predict(X_test)
rf_y_pred = rf_model.predict(X_test)
```

```python
# Evaluating and comparing the models

linear_score = linear_model.score(X_test, y_test)
print("Prediction Score for Linear Regression: ", linear_score)

svr_score = svr_model.score(X_test, y_test)
print("Prediction Score for SVR: ", svr_score)

rf_score = rf_model.score(X_test, y_test)
print("Prediction Score for Random Forest: ", rf_score)
```

```python
linear_r2_score = r2_score(y_test, linear_y_pred)
print("R-squared Score for Linear Regression: ", linear_r2_score)

svr_r2_score = r2_score(y_test, svr_y_pred)
print("R-squared Score for SVR: ", svr_r2_score)

rf_r2_score = r2_score(y_test, rf_y_pred)
print("R-squared Score for Random Forest: ", rf_r2_score)
```

```python
linear_mse = mean_squared_error(y_test, linear_y_pred)
print("Mean Squared Error for Linear Regression: ", linear_mse)

svr_mse = mean_squared_error(y_test, svr_y_pred)
print("Mean Squared Error for SVR: ", svr_mse)

rf_mse = mean_squared_error(y_test, rf_y_pred)
print("Mean Squared Error for Random Forest: ", rf_mse)
```
